```r
library('tidymodels')
```

```r
gss = read.csv('https://raw.githubusercontent.com/UNC-BIOS-512/jupyterlite/main/data/gss_subset.csv')

gss <- gss |> drop_na(realinc, educ)

gss |> glimpse()
```

```r
income_by_educ <- gss |>
    group_by(educ) |>
    summarize(realinc = mean(realinc))

income_by_educ |> head()
```

```r
mod=linear_reg()
mod
#Set up the linear regression model 
```

```r
mod_fit <- mod |> fit(realinc~educ, data=gss)
#modeling real income by education, using the gss data. 
mod_fit
```
